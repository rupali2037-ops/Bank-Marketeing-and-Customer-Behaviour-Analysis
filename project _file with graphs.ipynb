```python
import pandas as pd
import plotly.express as px

# Group by Class and count customers
conversion = (
    df.groupby("Class")
      .size()
      .reset_index(name="Customer Count")
)

# Replace numeric values with meaningful labels
conversion["Class"] = conversion["Class"].replace({
    1: "Subscribed",
    2: "Not Subscribed"
})

# Create Donut Chart
fig = px.pie(
    conversion,
    names="Class",
    values="Customer Count",
    hole=0.5,
    title="Conversion Rate Distribution",
    color="Class",
    color_discrete_map={
        "Subscribed": "green",
        "Not Subscribed": "red"
    }
)

fig.update_traces(
    textposition="inside",
    textinfo="label+percent+value"
)

fig.update_layout(
    title_x=0.5,
    legend_title="Customer Status"
)

fig.show()
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Create subscription labels
df["Subscription Status"] = df["Class"].replace({
    1: "Subscribed",
    2: "Not Subscribed"
})

# Group data
contact_subscription = (
    df.groupby(["Contact", "Subscription Status"])
      .size()
      .reset_index(name="Count")
)

# Get contact methods
contacts = contact_subscription["Contact"].unique()
fig = make_subplots(
    rows=1,
    cols=len(contacts),
    specs=[[{"type": "domain"}] * len(contacts)],
    subplot_titles=contacts
)

# Colors
colors = ["#003F88", "#8B0000"]  

# Add each pie chart
for i, contact in enumerate(contacts, start=1):
    data = contact_subscription[contact_subscription["Contact"] == contact]

    fig.add_trace(
        go.Pie(
            labels=data["Subscription Status"],
            values=data["Count"],
            pull=[0.1, 0.1],          
            marker=dict(
                colors=colors,
                line=dict(color="white", width=2)
            ),
            textinfo="percent+label",
            hole=0.1                   
        ),
        row=1,
        col=i
    )

fig.update_layout(
    title="Subscription Status by Contact Method (Exploded Pie Charts)",
    showlegend=True,
    width=2000,
    height=500
)

fig.show()
```

```python
import pandas as pd
import plotly.express as px

# Calculate subscription rate (%) by day
day_rate = (
    df.groupby("Contact day")["Class"]
      .apply(lambda x: (x == 1).mean() * 100)
      .reset_index(name="Subscription Rate")
)

# Sort days in ascending order
day_rate = day_rate.sort_values("Contact day")

# Line Chart
fig = px.line(
    day_rate,
    x="Contact day",
    y="Subscription Rate",
    markers=True,
    title="Subscription Rate by Day of Month"
)

fig.update_layout(
    xaxis_title="Day of Month",
    yaxis_title="Subscription Rate (%)"
)

fig.show()
```

```python
import plotly.express as px

# Create labels
df["Subscription Status"] = df["Class"].replace({
    1: "Subscribed",
    2: "Not Subscribed"
})

# Create pivot table
heatmap_data = pd.crosstab(
    df["Default"],
    df["Subscription Status"]
)

# Heatmap
fig = px.imshow(
    heatmap_data,
    text_auto=True,
    color_continuous_scale="Blues",
    title="Default Status vs Subscription Heatmap"
)

fig.update_layout(
    xaxis_title="Subscription Status",
    yaxis_title="Default Status"
)

fig.show()
```

```python
import plotly.express as px

# Create labels
df["Subscription Status"] = df["Class"].replace({
    1: "Subscribed",
    2: "Not Subscribed"
})

# Create heatmap data
heatmap_data = (
    df.groupby(["Subscription Status", "Previous contacts"])
      .size()
      .reset_index(name="Count")
)

# Heatmap
fig = px.density_heatmap(
    heatmap_data,
    x="Subscription Status",
    y="Previous contacts",
    z="Count",
    color_continuous_scale="reds",
    title="Heatmap of Previous Contacts by Subscription Status",
    text_auto=True
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    height=600,
    width=800,
    xaxis_title="Subscription Status",
    yaxis_title="Previous Contacts",
    coloraxis_colorbar_title="Customers"
)

fig.show()
```
